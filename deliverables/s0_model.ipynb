```python
mandates = {
    2025: 0.02,
    2030: 0.06,
    2035: 0.20,             # got the mandates from opening ceramony slides page 27
    2050: 0.70
}

def interpolate_mandates(year):
    years = sorted.keys(mandates)
    for i in range(len(years) - 1):
        if years[i] <= year <= years[i+1]:
            y0, y1 = years[i], years[i+1]
            s0,s1 = mandates[y0], mandates[y1]
            return s0 + (s1 - s0) * (year - y0) / (y1 - y0)
    return mandates[years[-1]]  # return last mandate if year is beyond last key
```
